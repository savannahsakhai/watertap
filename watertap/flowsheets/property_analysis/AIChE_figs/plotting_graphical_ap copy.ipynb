```python
data_MVC_seawater = pd.read_csv(r'C:\Users\sss0031\Documents\GitHub\watertap\watertap\flowsheets\property_analysis\tech_data\data_MVC_sea copy.csv')
data_MVC_reaktoro = pd.read_csv(r'C:\Users\sss0031\Documents\GitHub\watertap\watertap\flowsheets\property_analysis\tech_data\data_MVC_reaktoro.csv')
data_MVC_nacl = pd.read_csv(r'C:\Users\sss0031\Documents\GitHub\watertap\watertap\flowsheets\property_analysis\tech_data\data_MVC_nacl.csv')
data_MVC_simple = pd.read_csv(r'C:\Users\sss0031\Documents\GitHub\watertap\watertap\flowsheets\property_analysis\tech_data\data_MVC_simple.csv')
```

```python
sea_color = "#4A7B9D" 
nacl_color = "#467740"
reaktoro_color = "#0A0200"
```

```python
# Font sizes optimized for small figure
title_size = 6  # For subplot titles
label_size = 6  # For axis labels
tick_size = 6   # For tick labels

# Set global font parameters
plt.rcParams['font.size'] = tick_size
plt.rcParams['axes.labelsize'] = label_size
plt.rcParams['axes.titlesize'] = title_size
plt.rcParams['xtick.labelsize'] = tick_size
plt.rcParams['ytick.labelsize'] = tick_size
plt.rcParams['legend.fontsize'] = 6

fig, axs = plt.subplots(ncols=2, nrows=1, figsize=(3.22, 1.15), 
                        layout='constrained', dpi=450)

# Increase line width and marker sizes for visibility
n = 1.5  # Line width
marker_size = 3  # Larger markers

# Reverse osmosis plot
axs[0].plot(data_RO_reatoro['# Water Recovery']*100, data_RO_reatoro['LCOW'],
            "o", ms=marker_size+1, mfc='none', markevery=3, linewidth=n, 
            color=reaktoro_color, label="Reaktoro", zorder=3, markeredgewidth=0.5)
axs[0].plot(data_RO_seawater['# Water Recovery']*100, data_RO_seawater['LCOW'], 
            "*", ms=marker_size, markevery=3, linewidth=n, 
            color=sea_color, label="Lienhard", zorder=2, markeredgewidth=0.5)
axs[0].plot(data_RO_nacl['# Water Recovery']*100, data_RO_nacl['LCOW'], 
            "s", ms=marker_size, mfc='none', markevery=3, linewidth=n, 
            color=nacl_color, label="NaCl", zorder=1, markeredgewidth=0.5)

axs[0].set_xlabel("Water recovery (%)", fontsize=label_size, fontweight='bold', labelpad=1)
axs[0].set_ylabel("LCOW ($/m³)", fontsize=label_size, fontweight='bold', labelpad=2)
axs[0].set_title("Reverse osmosis", fontsize=title_size, pad=2, fontweight='bold')
axs[0].set_yticks(np.arange(0.6, 0.9, 0.1))
axs[0].set_xticks([30, 40, 50])
axs[0].tick_params(width=0.5, length=3, pad=1)
# axs[0].grid(True, alpha=0.3, linewidth=0.5)

# Mechanical vapor compression plot
axs[1].plot(data_MVC_reaktoro['# Water Recovery']*100, data_MVC_reaktoro['LCOW'], 
            "o", ms=marker_size+1, mfc='none', markevery=1, linewidth=n, 
            color=reaktoro_color, label="Reaktoro", zorder=3, markeredgewidth=0.5)
axs[1].plot(data_MVC_seawater['# Water Recovery']*100, data_MVC_seawater['LCOW'],
            "*", ms=marker_size, markevery=1, linewidth=n, 
            color=sea_color, label="Lienhard", zorder=2, markeredgewidth=0.5)
axs[1].plot(data_MVC_nacl['# Water Recovery']*100, data_MVC_nacl['LCOW'], 
            "s", ms=marker_size, mfc='none', markevery=1, linewidth=n, 
            color=nacl_color, label="NaCl", zorder=1, markeredgewidth=0.5)

axs[1].set_xlabel("Water recovery (%)", fontsize=label_size, fontweight='bold', labelpad=1)
axs[1].set_ylabel("LCOW ($/m³)", fontsize=label_size, fontweight='bold', labelpad=2)
axs[1].set_title("Mechanical vapor compression", fontsize=title_size, pad=2, fontweight='bold')
axs[1].set_yticks(np.arange(4.3, 4.65, 0.1))
axs[1].set_xticks([40, 55, 70])
axs[1].tick_params(width=0.5, length=3, pad=1)
# axs[1].grid(True, alpha=0.3, linewidth=0.5)

# # Add legend to one subplot (top right corner, small)
# axs[1].legend(loc='upper right', frameon=True, framealpha=0.9, 
#               edgecolor='black', fancybox=False, fontsize=5)

# Adjust spine width for better visibility
for ax in axs:
    for spine in ax.spines.values():
        spine.set_linewidth(0.5)

plt.savefig('graphical_abstract.png', dpi=450, bbox_inches='tight', 
            facecolor='white', pad_inches=0.02)
plt.savefig('graphical_abstract.tif', dpi=450, bbox_inches='tight',
            facecolor='white', pad_inches=0.02)
```
